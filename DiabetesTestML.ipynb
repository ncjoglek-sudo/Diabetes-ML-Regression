```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import HistGradientBoostingRegressor
import joblib
```

```python
#import data
df = pd.read_csv('clean_diabetes.csv')
#identify variables and dependencies, pregnancies, skinthickness, and bloodpressure excluded
X = df[['glucose','insulin','bmi','diabetespedigreefunction','age']]
y = df['outcome']
#train data parameters
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
#train regression data
model = HistGradientBoostingRegressor(max_iter=100, learning_rate=0.1, random_state=42)
#fit model
model.fit(X_train, y_train)
```

```python
#predict
#y_pred = model.predict(X_test)
y_pred = model.predict(X_test)
#
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
#
print("MSE:", mse)
print("R2:", r2)
```

```python
from sklearn.inspection import permutation_importance
#
result = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
#
for feature, importance in zip(X.columns, result.importances_mean):
    print(feature, importance)
```

```python
#compare training and testing performance
train_r2 = model.score(X_train, y_train)
test_r2 = model.score(X_test, y_test)
#
print("Train R²:", train_r2)
print("Test R²:", test_r2)
```

```python
#finish model 
joblib.dump(model, "my_model.pkl")
```
